```python
# TODO: اندازه تصاویر و اندازه هر بچ (تعداد تصاویر در هر دسته آموزش) را مشخص کنید
import tensorflow as tf


# تعیین اندازه تصاویر و Batch
IMG_SIZE = (200, 300)
BATCH_SIZE = 20

# بارگذاری داده‌های آموزش
train_ds = tf.keras.utils.image_dataset_from_directory(
    BASE_DIR,
    validation_split=0.2,
    subset='training',
    seed=42,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='categorical'
)

# بارگذاری داده‌های اعتبارسنجی
val_ds = tf.keras.utils.image_dataset_from_directory(
    BASE_DIR,
    validation_split=0.2,
    subset='validation',
    seed=42,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='categorical'
)

# نمایش نام کلاس‌ها
print("Class Names:", train_ds.class_names)

# دریافت یک Batch
images, labels = next(iter(train_ds))

# نمایش ابعاد تصاویر و برچسب‌ها
print("Images shape:", images.shape)
print("Labels shape:", labels.shape)


```

  <font size=4>
    <li>
        این متد یک batch از تصاویر دیتاست را دریافت کرده، لایه‌ی  augmentation را روی آن‌ها اعمال می‌کند و تعدادی از تصاویر را در قالب یک شبکه تصویری نمایش می‌دهد. این کار برای بررسی عملکرد لایه‌ی augmentation و اطمینان از درست بودن اعمال آن روی تصاویر انجام می‌شود. متد زیر را کامل کنید.
    </li>

<br>

```python

import tensorflow as tf
import matplotlib.pyplot as plt
import math

def visualize_augmented_images(dataset, augmentation_layer, num_samples=9):

    # دریافت یک Batch از دیتاست
    images, labels = next(iter(dataset))

    # بررسی تعداد تصاویر در Batch
    if num_samples > images.shape[0]:
        raise ValueError("num_samples cannot exceed batch size")

    # اعمال Data Augmentation
    augmented_images = augmentation_layer(images, training=True)

    # تعیین تعداد ردیف و ستون
    cols = math.ceil(math.sqrt(num_samples))
    rows = math.ceil(num_samples / cols)

    # ایجاد محیط نمایش
    plt.figure(figsize=(10, 10))

    # نمایش تصاویر
    for i in range(num_samples):

        plt.subplot(rows, cols, i + 1)

        image = tf.clip_by_value(
            augmented_images[i], 0, 255
        )

        plt.imshow(image.numpy().astype("uint8"))

        plt.axis('off')

    plt.tight_layout()
    plt.show()
```

<br>
<div dir='rtl'>
  <font size=4>
    <li>
        این بخش از باید پیش‌پردازش (Preprocessing) روی داده‌های تصویری قبل از ورود به مدل انجام دهید.
         در اینجا هدف اصلی نرمال‌سازی مقادیر پیکسل‌ها است، به این صورت که مقادیر اصلی که در بازه [0, 255] هستند، به بازه [0.0, 1.0] تبدیل می‌شوند.
         یک تابع تعریف شده که هر تصویر را از نوع int به float تبدیل می‌کند و سپس آن را بر ۲۵۵ تقسیم می‌کند تا نرمال شود. این تابع را کامل کنید و سپس  روی کل داده‌های آموزش و اعتبارسنجی با استفاده از متد map اعمال شده است.
    </li>
  </font>
<br><br>

```python

def preprocess(image, label):

    # تبدیل نوع داده از int به float
    image = tf.cast(image, tf.float32)

    # نرمال‌سازی مقادیر پیکسل
    image = image / 255.0

    return image, label

```

```python

import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.Input(shape=(200, 300, 3)),

    # Data Augmentation
    data_augmentation,

    # Block 1
    tf.keras.layers.Conv2D(32, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.Conv2D(32, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.MaxPooling2D(2),

    # Block 2
    tf.keras.layers.Conv2D(64, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.Conv2D(64, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.MaxPooling2D(2),

    # Block 3
    tf.keras.layers.Conv2D(128, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.Conv2D(128, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.MaxPooling2D(2),

    # Block 4
    tf.keras.layers.Conv2D(256, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.Conv2D(256, 3, padding='same', use_bias=False),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Activation('relu'),

    tf.keras.layers.MaxPooling2D(2),

    # Classification
    tf.keras.layers.GlobalAveragePooling2D(),

    tf.keras.layers.Dense(256, activation='relu'),

    tf.keras.layers.Dropout(0.5),

    tf.keras.layers.Dense(3, activation='softmax')
])

# Compile
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0003
    ),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python

import matplotlib.pyplot as plt

def plot_training_history(history):

    # دریافت اطلاعات دقت و خطا
    acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']

    loss = history.history['loss']
    val_loss = history.history['val_loss']

    # تعداد Epoch ها
    epochs = range(1, len(acc) + 1)

    # ایجاد محیط نمایش دو نمودار
    plt.figure(figsize=(14, 5))

    # نمودار Accuracy
    plt.subplot(1, 2, 1)

    plt.plot(epochs, acc, label='Training Accuracy')
    plt.plot(epochs, val_acc, label='Validation Accuracy')

    plt.title('Training and Validation Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    # نمودار Loss
    plt.subplot(1, 2, 2)

    plt.plot(epochs, loss, label='Training Loss')
    plt.plot(epochs, val_loss, label='Validation Loss')

    plt.title('Training and Validation Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def predict_image(model, image_path, class_names, img_size=(180, 180)):

    # 1. بارگذاری تصویر
    image = Image.open(image_path).convert('RGB')

    # 2. تغییر اندازه تصویر
    image = image.resize(img_size)

    # 3. تبدیل تصویر به آرایه NumPy
    image_array = np.array(image)

    # 4. نرمال‌سازی تصویر
    image_array = image_array.astype('float32') / 255.0

    # 5. اضافه کردن بعد Batch
    image_array = np.expand_dims(image_array, axis=0)

    # 6. پیش‌بینی با مدل
    predictions = model.predict(image_array)

    # 7. پیدا کردن کلاس با بیشترین احتمال
    predicted_index = np.argmax(predictions[0])

    predicted_class = class_names[predicted_index]

    # 8. محاسبه میزان اطمینان
    confidence = predictions[0][predicted_index] * 100

    # 9. نمایش تصویر و نتیجه
    plt.figure(figsize=(5, 5))
    plt.imshow(image)
    plt.title(f'{predicted_class} ({confidence:.2f}%)')
    plt.axis('off')
    plt.show()

    return predicted_class, confidence
```
